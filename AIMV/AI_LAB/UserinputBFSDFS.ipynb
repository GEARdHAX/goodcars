```python
# -----------------------------------
# Install libraries (run once)
# -----------------------------------
!pip install networkx matplotlib

# -----------------------------------
# Import libraries
# -----------------------------------
import networkx as nx
import matplotlib.pyplot as plt
from collections import deque

# -----------------------------------
# Take graph input from user
# -----------------------------------
G = nx.Graph()

n = int(input("Enter number of edges: "))

print("Enter edges (node1 node2):")
for _ in range(n):
    u, v = input().split()
    G.add_edge(u, v)

start = input("Enter start node: ")
goal = input("Enter goal node: ")

pos = nx.spring_layout(G)

# -----------------------------------
# Draw function
# -----------------------------------
def draw_graph(visited, current):
    plt.clf()

    node_colors = []
    for node in G.nodes():
        if node == current:
            node_colors.append('red')   # current node
        elif node in visited:
            node_colors.append('green') # visited
        else:
            node_colors.append('lightblue')

    nx.draw(G, pos, with_labels=True, node_color=node_colors, node_size=3000)
    plt.title(f"BFS Visiting: {current}")
    plt.pause(1)

# -----------------------------------
# BFS Algorithm
# -----------------------------------
def bfs():
    OPEN = deque([start])
    visited = set([start])
    CLOSE = []

    plt.figure()

    while OPEN:
        N = OPEN.popleft()
        CLOSE.append(N)

        draw_graph(CLOSE, N)

        if N == goal:
            print("\nGoal Found!")
            print("Traversal:", CLOSE)
            break

        for neighbor in G[N]:
            if neighbor not in visited:
                OPEN.append(neighbor)
                visited.add(neighbor)

    plt.show()

# Run BFS
bfs()
```
